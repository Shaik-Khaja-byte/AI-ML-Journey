## Different plot functions in matplotlib
### Bar chart - used to represent categorical data can be horizontal or vertical

```python
oscar_movies = [
    "Birdman",
    "Spotlight",
    "Moonlight",
    "Green Book",
    "Parasite"
]

years = [2015, 2016, 2017, 2018, 2019]

oscar_revenue = [103, 98, 65, 321, 263]

plt.bar(years, oscar_revenue, color="purple")
plt.title("Year wise revenue for Oscar Movies")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")

# plt.text(x_axis_pos, y_axis_pos, value, alignment) - suppse you want to add a label to each bar (the exact value sometimes to make the data look clear)

for i in range(len(years)):
    plt.text(years[i], oscar_revenue[i]+7, str(oscar_revenue[i]), ha="center")

# plt.ylim(lower, upper), plt.xlimit(hor lower, hor upper)
plt.ylim(0, max(oscar_revenue)+40)

plt.show()
```

## comparing the data using bar charts

```python
import numpy as np

non_oscar_revenue = [1005, 839, 731, 214, 101]
oscar_revenue = [103, 98, 65, 321, 263]

years = [2008, 2010, 2014, 1994, 1999]

x  = np.arange(len(years))
width = 0.4

plt.bar(x-width/2, oscar_revenue, width, label="Oscar") # moving to the LEFT
plt.bar(x+width/2, non_oscar_revenue, width, label="non-Oscar") # moving to the RIGHT

plt.title("Oscar v/s non-Oscar movies revenue")
plt.xlabel("Years")
plt.ylabel("Revenue ($M)")
plt.legend()

plt.xticks(x, years)

plt.show()
```

```python
# changing color based on some condition : safe or high risk

people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]

blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

color = ["green" if x < 135 else "red" for x in blood_pressure]

plt.scatter(age, blood_pressure, color=color)
plt.title("Age v/s BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
plt.show()
```

## Annotations - exactly annotate/tell the dot/marker belongs to whom
### plt.annotate("text", xy=(x_point, y_point), xytext=(x_point+-n, y_point+-n))

```python
# annotate()

people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]

age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]

blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

plt.scatter(age, blood_pressure, cmap="OrRd", c=blood_pressure, s=blood_pressure)
plt.title("Age v/s BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.colorbar(label="BP")
plt.grid()

for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i], blood_pressure[i]), xytext=(age[i]+1, blood_pressure[i]+1))

plt.xlim(min(age), max(age)+10)
plt.ylim(min(blood_pressure), max(blood_pressure)+5)
plt.show()
```
