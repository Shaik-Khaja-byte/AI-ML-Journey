# Matplotlib - standard library for data visualisation

```python
import matplotlib.pyplot as plt # pyplot is the drawing tool in matplotlib (which contains a lot of functions)
```

```python
x = [1, 2, 3, 4]
y = [4, 5, 6, 7]

plt.plot(x, y) # plot() - line plot
plt.show()
```

```python
# Oscar Winner Movies
oscar_movies = [
    "Birdman",
    "Spotlight",
    "Moonlight",
    "Green Book",
    "Parasite"
]

years = [2015, 2016, 2017, 2018, 2019]

oscar_revenue = [103, 98, 65, 321, 263]

plt.plot(years, oscar_revenue)

plt.title("Oscar Winning Movies Revenue in each year")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.show()
```

```python
# Non-Oscar Winner Movies
non_oscar_movies = [
    "The Dark Knight",
    "Inception",
    "Interstellar",
    "Pulp Fiction",
    "Fight Club"
]

years = [2008, 2010, 2014, 1994, 1999]

non_oscar_revenue = [1005, 839, 731, 214, 101]

plt.plot(years, oscar_revenue, label="Oscar")
plt.plot(years, non_oscar_revenue, label="non-Oscar")

plt.title("oscar v/s non oscar movies revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.show()
```

## format strings - used to change the style of the lines in plots
- it consists of a part for marker, line and color

```python
# Non-Oscar Winner Movies
non_oscar_movies = [
    "The Dark Knight",
    "Inception",
    "Interstellar",
    "Pulp Fiction",
    "Fight Club"
]

years = [2008, 2010, 2014, 1994, 1999]

non_oscar_revenue = [1005, 839, 731, 214, 101]

plt.plot(years, oscar_revenue, color="g", marker="o", linestyle="-.", linewidth="3" ,label="Oscar")
plt.plot(years, non_oscar_revenue, "^--r" ,label="non-Oscar")

plt.title("oscar v/s non oscar movies revenue")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.show()
```

```python
# tight_layout() - improves the label congestion


plt.style.use("default")

movies = [
    "The Dark Knight",
    "Inception",
    "Interstellar",
    "Pulp Fiction",
    "Fight Club"
]

revenue = [1005, 839, 731, 214, 101]

plt.plot(movies, revenue)

plt.title("Movies Revenue")
plt.xlabel("Movies")
plt.ylabel("Revenue (in $M)")
plt.tight_layout()

# savefig("abc.png") - you can download your plot

plt.savefig("Tight Layout Plot.png")

plt.show() # always use show in the end
```

## Different plot functions in matplotlib
### Bar chart - used to represent categorical data can be horizontal or vertical

```python
oscar_movies = [
    "Birdman",
    "Spotlight",
    "Moonlight",
    "Green Book",
    "Parasite"
]

years = [2015, 2016, 2017, 2018, 2019]

oscar_revenue = [103, 98, 65, 321, 263]

plt.bar(years, oscar_revenue, color="purple")
plt.title("Year wise revenue for Oscar Movies")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")

# plt.text(x_axis_pos, y_axis_pos, value, alignment) - suppse you want to add a label to each bar (the exact value sometimes to make the data look clear)

for i in range(len(years)):
    plt.text(years[i], oscar_revenue[i]+7, str(oscar_revenue[i]), ha="center")

# plt.ylim(lower, upper), plt.xlimit(hor lower, hor upper)
plt.ylim(0, max(oscar_revenue)+40)

plt.show()
```

## comparing the data using bar charts

```python
import numpy as np

non_oscar_movies = [
    "The Dark Knight",
    "Inception",
    "Interstellar",
    "Pulp Fiction",
    "Fight Club"
]

non_oscar_movies = [
    "The Dark Knight",
    "Inception",
    "Interstellar",
    "Pulp Fiction",
    "Fight Club"
]


non_oscar_revenue = [1005, 839, 731, 214, 101]
oscar_revenue = [103, 98, 65, 321, 263]

years = [2008, 2010, 2014, 1994, 1999]

x  = np.arange(len(years))
width = 0.4

plt.bar(x-width/2, oscar_revenue, width, label="Oscar") # moving to the LEFT
plt.bar(x+width/2, non_oscar_revenue, width, label="non-Oscar") # moving to the RIGHT

plt.title("Oscar v/s non-Oscar movies revenue")
plt.xlabel("Years")
plt.ylabel("Revenue ($M)")
plt.legend()

plt.xticks(x, years)

plt.show()
```
