# NumPy

## designed for fast numerical computing

### python lists - heterogeneous, slow
### numpy arrays (ndarray) - homogeneous, fast because it is optimised with C language 

```python
import numpy as np
```

```python
arr = np.array([1, 2, 3, 4, 5])
print(arr, type(arr))
```

#### how fast are numpy arrays?

```python
import time

size = 100_000_000

py_list = list(range(size))
start = time.time()
sq_list = [x**2 for x in py_list]
end = time.time()
print(f"Python List : {end - start} seconds")

np_arr = np.array(py_list)
start = time.time()
sq_arr = np_arr ** 2 # notice that we didn't use the loop at all, we just performed the operation as a whole on the array (VECTORIZATION)
end = time.time()
print(f"Numpy Array : {end - start} seconds")
```

#### let's compare memory aswell?

```python
import sys

print(f"Python List : {sys.getsizeof(py_list) * len(py_list)} bytes")
print(f"Numpy Array : {np_arr.nbytes} bytes")
```

#### why python lists occupy more memory?
- numpy arrays store elements in contiguous memory location
- whereas the list stores each element in different location, with different location comes the meta-data associated with it (address and other stuff)
