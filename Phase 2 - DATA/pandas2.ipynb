# Data Transformation - Feature engineering

### it is the process of adding a new feature/column so that the accuracy of the model increases

```python
import pandas as pd 
df = pd.read_csv("data files/raw_data.csv")
df
```

```python
df.columns
```

### apply() - it is a higher order function which takes another function as parameter to perform functions like adding a new column

```python
df2 = df.copy()

df2["tax"] = df2["income"].apply(lambda x : "20%" if x >= 60000 else "10%")
df2
```

### map() - changing the existing values and mapping it with the new ones

```python
# change the values of: Male -> M, Female -> F, Unknown -> U

gender_map = {"Male" : "M", "Female" : "F", "Unknown" : "U"}

df2["gender"] = df2["gender"].map(gender_map)

df2
```

```python
# assign() - creating new columns by assigning them some values
# example: increase the salary by 10%

df2.assign(new_income = df["income"] * 1.1)
```

```python
# replace(old, new) - used to replace the old value from the new

df2["country"] = df2["country"].replace("USA", "United States") 
df2
```

```python
# rename(columns = {old : new}) - rename rows and columns
# if you want to rename all the colums:

df2.columns = ["Id", "Name", "Age", "Country", "Gender", "Income", "Tax"]
df2
```

```python
# but if you want to rename a single column?

df2.rename(columns= {"Income" : "Salary"}) # renaming columns
```

```python
df2.rename(index= {1 : "First"}) # renaming row
```

```python
# sorted_values() - sorts a column or a set of columns

df2 = df2.fillna(50)
df2
```

```python
df2.sort_values("Income") # sort based on income
```

```python
sorted_df = df2.sort_values(["Income", "Age"]) # sort based on income and age
sorted_df
```

```python
df2.sort_values("Income", ascending=False) # sort in descending order
```

```python
sorted_df.sort_index() # sort based on index values
```

```python
sorted_df.reset_index() # the data is sorted, index is preserved and it has added new labels
```

```python
sorted_df.reset_index(drop= True) # removes preserved indexes
```

```python
sorted_df["Ranking"] = sorted_df["Income"].rank()
sorted_df
# you can observe, the 5.5 which is a bit sus, but thats the average b/w 5 and 6 
# and for the tie rows, it will be the same i.e., 5.5
```

```python
sorted_df["Ranking"] = sorted_df["Income"].rank(ascending= False)
sorted_df
```

### there are several options on how you want to resove tie of the ranks, you can specify a param in the rank function i.e., rank(method= dense, min, max)


```python
sorted_df["Ranking"] = sorted_df["Income"].rank(method= "dense") # tie -> same number
sorted_df
```

```python
sorted_df["Ranking"] = sorted_df["Income"].rank(method= "min") # tie -> min will be repaeted
sorted_df
```

```python
sorted_df["Ranking"] = sorted_df["Income"].rank(method= "max") # tie -> max will be repeated
sorted_df
```

### if you want to calculate mulyiple things like mean, max, min at the same time - agg()

```python
df.groupby("country")["income"].agg(["mean", "max", "min"])
```

```python
df.groupby("country")["income"].aggregate(["mean", "max", "min"]) # aggregate is an alias of agg
```

```python
# but you can change the column names in agg
df.groupby("country")["income"].agg(avg_salary = "mean", max_salary = "max", min_salary = "min")
```

```python
# you can groupby any column with any function

df.groupby("country").agg({
    "income" : "mean",
    "age" : "max"
})
```

```python
# if you want to change the column names
df.groupby("country").agg(
    avg_income= ("income" , "mean"),
    max_age= ("age" , "max")
)
```

```python
df = pd.DataFrame({
    "country": ["USA", "USA", "India", "India"],
    "year": [2020, 2021, 2020, 2021],
    "sales": [100, 120, 90, 110],
    "profit": [20, 25, 18, 22]
})

df
```

```python
# using the melt(id_vars, value_vars, var_name, value_name) - [wide to long] to melt the table and make it long

melt_df = df.melt(
    id_vars=["country", "year"],
    value_vars=["sales", "profit"],
    var_name="metrics",
    value_name="value"
)

melt_df
```

```python
# using the pivot(index, columns, values) - [long to wide] used to convert melted data into normal data

original = melt_df.pivot(
    index=["country", "year"],
    columns="metrics",
    values="value"
)

original
```

## Joins

```python
df_customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Adam", "Bob", "Charlie", "Dave"]
})

df_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [2, 1, 4, 5],
    "amount": [250, 120, 300, 180]
})
```

```python
# inner join - only matching values from both the tables

pd.merge(df_customers, df_orders, on="customer_id")
```

```python
# left join - all the values from left and replace non matcing with NaN in right table

pd.merge(df_customers, df_orders, on="customer_id", how="left")
```

```python
# right join - all the values from right and replace non matcing with NaN in left table

pd.merge(df_customers, df_orders, on="customer_id", how="right")
```

```python
# outer join - combines all the values in two tables

pd.merge(df_customers, df_orders, on="customer_id", how="outer")
```

## Data Conactenation - stacing dataframes on top of each other (row) or side by side (column)

```python
df1 = pd.DataFrame({
    "id" : [1,2,3],
    "name" : ["shaik", "khaja", "rahul"]
})

df2 = pd.DataFrame({
    "id" : [4,5,6],
    "name" : ["shreyas", "venkat", "saraswath"]
})
```

```python
# row concat
pd.concat([df1, df2])
```

```python
# column concat
pd.concat([df1, df2], axis=1)
```
