## Series in pandas
### 1D labelled arrays

```python
s = pd.Series([1,2,3,4,5])

print(s)
print(type(s))

# index - positions
print(s[2])
print(s[3])
```

```python
# labels = named index

s = pd.Series([22, 23, 24, 25], index = ["ajay", "shreyas", "saraswath", "nick"])

print(s)
print(s["ajay"])
print(s["nick"])
print(s.index)
```

## functions associated with series
- Homogeneous - holds same values like numpy
- vectorized operations
- Handles missing values with NaN
- Mutable values, Immutable size

```python
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([10, 20, 30, 40, 50])

print(s1 + s2)

s1[0] = 100

changed_s1 = s1.drop(0) # original s1 is still intact - immutable size
print(s1)
print(changed_s1)
```

## Filtering of Data

```python
df[ df["aqi"] > 100]

df[ (df["aqi"] > 100) & (df["temperature"] > 30)]
df[ (df["aqi"] > 100) & (df["temperature"] < 30)] # you can observe a small pattern over here by looking at the no. of rows

aqi_data = df[ (df["aqi"] > 100) & (df["temperature"] > 30)] [["city", "aqi"]] # specifying separate colums in [[]] brackets

aqi_data.iloc[0]
# aqi_data.loc[0] label 0 is not available
aqi_data.loc[6]
```

```python
df
```

## query method - a simple way of filtering

```python
df.query("aqi > 100")
```

```python
df.query("aqi > 150 & temperature > 30") [["city", "aqi"]]
```

```python
aqi_var = 160
df.query("aqi > @aqi_var & temperature > 30") [["city", "aqi"]] # you can also use a local variable in the query using @
```

## Data Cleaning

- isnull()

```python
df = pd.read_csv("data files/raw_data.csv")
df
```

```python
df.isnull() #  returns true if it detects NaN (null)
df.isna() # alias of isnull() gives the same result, use any one
```

```python
df.isnull().sum() # you can chain the functions
```

```python
df.dropna() # drops the rows with the null values
```

```python
df.dropna(axis= 1) # drops the colums with the null values
```

```python
df.fillna(0) # fills the null values with the value specified
df.fillna(100)
```

```python
cleaned_data = df.copy()
age_mean = cleaned_data["age"].mean()

cleaned_data["age"] = cleaned_data["age"].fillna(age_mean)
cleaned_data
```

```python
df
```

```python
df.ffill() # forward fill - top to botton repeat the top value if next is NaN
```

```python
df.bfill() # backward fill - bottom to top repeat the bottom value if next is NaN
```

```python
df
```

## Handle Duplicates

```python
# duplicated() - checks if duplicates are present

df.duplicated()

# you can check for a single column as well
df["country"].duplicated()

# combination of columns also you can check
df[["country", "gender"]].duplicated()
```

```python
# drop_duplicates() - drops the duplicate rows

df.drop_duplicates() # labels are preserved
```

## Handling Data Types and date-time

```python
# dtypes - returns the datatypes of all the columns

df.dtypes
```

```python
# astype(datatype) - changes the data to our desired type

df2 = df.copy()

df2 = df2.fillna(0)
df2["age"] = df2["age"].astype("int64").copy()

df2.dtypes
```

```python
df2
```

```python
# to_datetime() -> converts the type to datetime datatype

date_str = pd.Series(["2025-08-25"])
type(date_str.dtypes)
```

```python
date_str = pd.Series([pd.to_datetime("2025-08-25")])
type(date_str.dtypes)
```
