## Series in pandas
### 1D labelled arrays

```python
s = pd.Series([1,2,3,4,5])

print(s)
print(type(s))

# index - positions
print(s[2])
print(s[3])
```

```python
# labels = named index

s = pd.Series([22, 23, 24, 25], index = ["ajay", "shreyas", "saraswath", "nick"])

print(s)
print(s["ajay"])
print(s["nick"])
print(s.index)
```

## functions associated with series
- Homogeneous - holds same values like numpy
- vectorized operations
- Handles missing values with NaN
- Mutable values, Immutable size

```python
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([10, 20, 30, 40, 50])

print(s1 + s2)

s1[0] = 100

changed_s1 = s1.drop(0) # original s1 is still intact - immutable size
print(s1)
print(changed_s1)
```
