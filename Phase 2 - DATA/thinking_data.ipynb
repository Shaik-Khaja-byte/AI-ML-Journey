# Assignment 1
## cleaning and structuring our data
- inconsistent data (ratings as string and number)
- missing data (age)
- duplicate data

```python
# Clean the data
def clean_data(data):
    # as the ratings are from 1 to 5 we can create a map
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    cleaned_data = []
    unique_users = set()

    # handling inconsistencies
    for user in data:
        raw_ratings = str(user["rating"]).strip().lower()
        if raw_ratings in text_to_num:
            raw_ratings = text_to_num[raw_ratings]
        user["rating"] = raw_ratings

        # handling missing data
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None

        # handling duplicates - Deduplication
        if user["name"].strip() in unique_users:
            continue

        unique_users.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data
        
```

```python
data = clean_data(data)
print(data)
```

```python
# get meaningful insights

def get_insights(data):

    # calculating the average rating
    tot_rating = 0

    for user in data:
        tot_rating += float(user["rating"])

    print(f"avg rating = {tot_rating / len(data)}")

    poor_ratings = 0
    for user in data:
        if float(user["rating"]) < 3:
            poor_ratings += 1

    print(f"% of users with poor ratings : {poor_ratings/len(data) * 100} %")
```

```python
get_insights(data)
```

```python
# building a recomendataion system

def get_recomendations(data):
    recomendations = []

    for user in data:
        curr_recomm = {}
        curr_recomm["name"] = user["name"]

        if(float(user["rating"]) >= 4):
            curr_recomm["brand"] = "apple"
        else:
            curr_recomm["brand"] = "samsung"

        recomendations.append(curr_recomm)
    return recomendations
```

```python
get_recomendations(data)
```
