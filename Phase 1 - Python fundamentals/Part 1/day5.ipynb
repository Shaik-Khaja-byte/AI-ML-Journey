```python
# open() function - to open a file in order to perform other operations

f = open("hello.txt", "r") # returns a file-object

data = f.read() # after opening the file, to read the data read() function is used
print(data)

f.close() # once you open a file, you should close it explicitly (data is valuable, hence do not leave it open)
```

```python
f = open("hello.txt", "r") 

data = f.readline() # reads the file line by line
print(data)

data = f.readline() 
print(data)
```

```python
f = open("hello.txt", "w") # you cannot write the file in read mode (common sense)

data = f.write("He actually resigned \nbut he says, he resigned because of \nhis personal reasons") # write overwrites the entire document

print(data) # write retruns the number of characters sucessfully written to the file, that's why u see a number there
```

```python
# different modes associated with files
# we have seen read(r) and write(w) modes
```

```python
# append(a) mode, it starts addiing the text from the end of the existing file content
f = open("hello.txt", "a")
f.write("Hey, I'm appending the text")
# after running check the file

# if the existing files do not exists... w and a modes create one for writing
```
