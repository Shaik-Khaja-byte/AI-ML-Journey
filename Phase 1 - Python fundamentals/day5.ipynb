```python
# open() function - to open a file in order to perform other operations

f = open("hello.txt", "r") # returns a file-object

data = f.read() # after opening the file, to read the data read() function is used
print(data)

f.close() # once you open a file, you should close it explicitly (data is valuable, hence do not leave it open)
```

```python
f = open("hello.txt", "w") # you cannot write the file in read mode (common sense)

data = f.write("He actually resigned \nbut he says, he resigned because of \nhis personal reasons") # write overwrites the entire document

print(data) # write retruns the number of characters sucessfully written to the file, that's why u see a number there
```

```python
# different modes associated with files
# we have seen read(r) and write(w) modes
```

```python
# append(a) mode, it starts addiing the text from the end of the existing file content
f = open("hello.txt", "a")
f.write("Hey, I'm appending the text")
# after running check the file

# if the existing files do not exists... w and a modes create one for writing
```

```python
# x - dedicatedly creates a new file and opens for writing
f = open("hello2.txt", "x")
f.write("some random text")
f.close()
```

```python
# b - binary mode is also available
# you can combine different modes too like :
# rt - read mode with text
# rb - read binary data
# similarly wt, wb
# but remember BY DEFAULT if you dont specify anything with r - means it's for text mode only
# + - opens disk file for update (r & w) for ex:
# r+ - you can perform read but theres an extra + so you write too... it can be used with w+, a+ etc
```

```python
f = open("hello.txt", "r+")
f.write("This is r+ mode")
print(f.read()) # because the pointer after writing is at igned so from there it is printing
```

```python
f = open("hello.txt", "a+")
f.write("This is a+ mode")
print(f.read()) # it's not printing anything cuz after writing, from the pointer's position it will start to print but there's nothing after writing is completed
```

```python
# activity solution
data = True
word = "python"
line = 1

with open("example.txt", "r") as f:
    while data:
        data = f.readline()

        if word in data:
            print(f"{word} found at line {line}")

        line += 1
```

```python
# exception handling - some errors can disrupt the flow of the program, there are some errors which we can predict
# handling those errors(anomalies) is called exception handling
# try, except, else, finally
```

```python
try: # code which might throw an exception is written in try block
    x = int(input("Enter an integer: "))
    print(f"you entered {x}")
    ans = 10/x

except ZeroDivisionError: # except catches the exception and handles - there are many built-in exceptions and you can create your own except blocks as well
    print("Division by zero is not defined.")

except ValueError: # you can list as many except blocks as you want
    print("The data entered is not integer.")

else: # if theres no exception, this statement will execute
    print("ans = ", ans)

finally: # exceutes the statements irrespective of the error thrown or not
    print("end of the program")
```

```python
# list comprehension - a compact way of doing the things 
# there is a fixed format of doing it -> [output | for item in range | condition] (condition is optional)
```

```python
# let me show you the traditional way v/s the list comprehension
# print an array with the squares from 1-5

squares = []

for i in range(1, 6):
    squares.append(i*i)

print(squares)
```

```python
# using list comprehension

sq = [i*i for i in range(1, 6)] # we have skipped the condition as it was not needed
print(sq)
```

```python
# now we have a condition : I need squares of only odd numbers :)

sq = [i*i for i in range(1, 6) if i%2 != 0]
print(sq)
```

```python
# task - given a list [-1, -2, 1, 2, 3, -1]
# output - [0, 0, 1, 2, 3, 0]

list = [-1, -2, 1, 2, 3, -1]

list = [0 if val<0 else val for val in list]
print(list)
```

```python
# task - convert the list to uppercase - ["slay", "the", "day"]
# output - ['SLAY', 'THE', 'DAY']

list = ["slay", "the", "day"]

list = [item.upper() for item in list]
print(list)
```

```python
# json module -> if you are working with development team, and you need the data to create a model, they'll give you the data in json format
# you should be able to convert it from json to python dict to make it work in python
# json.loads() - json_string -> python_object
# json.dumps() - python_object -> json_string
# take a moment in identifying the changes in the data types when you convert from json_str -> py_obj and vice versa
```

```python
import json

json_str = '{"name": "Shaik Khaja", "isStudent": true}'
py_obj = json.loads(json_str)

print(type(py_obj), py_obj)
```

```python
import json

py_obj = {"name": "Shaik Khaja", "isStudent": None}
json_str = json.dumps(py_obj)

print(type(json_str), json_str)
```

```python
# Write 5 names to a file and read them back

with open("ass1.txt", "w") as f:
    for i in range(5):
        name = input(f"enter the name {i+1}: ")
        f.write(name + "\n")

data = True

with open("ass1.txt", "r") as f:
    while data:
        data = f.readline()
        print(data)
```

```python
# Create, save, load, update a JSON file
import json

cities = {
    "Delhi": 32000000,
    "Mumbai": 21000000,
    "Bengaluru": 14000000
}

with open("cities.json", "w") as f:
    json.dump(cities, f, indent=4)


with open("cities.json", "r") as f:
    data = json.load(f)

new_city = input()
new_population = int(input())

data[new_city] = new_population

with open("cities.json", "w") as f:
    json.dump(data, f, indent=4)

print("city added successfully")
```
